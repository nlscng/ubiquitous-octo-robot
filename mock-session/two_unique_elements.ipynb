This problem was asked by Facebook.

Given an array of integers in which two elements appear exactly once and all other elements appear exactly twice, find the two elements that appear only once.

For example, given the array [2, 4, 6, 8, 10, 2, 6, 10], return 4 and 8. The order does not matter.

Follow-up: Can you do this in linear time and constant space?

```python

from typing import List, Optional, Tuple
```

```python
def two_uniques(arr: List[int]) -> Tuple[int, int]:
    assert arr and len(arr) >= 2, "Input array must have at least two elements."
    coll = set()
    for num in arr:
        if num in coll:
            coll.remove(num)
        else:
            coll.add(num)
    if len(coll) != 2:
        raise ValueError("Input array must contain exactly two unique elements.")
    else:
        return tuple(coll)
```

```python
#### Test cases
assert two_uniques([1, 2, 3, 2, 1, 4]) == (3, 4)
assert two_uniques([5, 6, 7, 5, 6, 8]) == (7, 8), "Test case with different numbers failed. Expected (7, 8), contains: " + str(two_uniques([5, 6, 7, 5, 6, 8]))
assert two_uniques([9, 10, 11, 9, 10, 12]) == (11, 12)
```
